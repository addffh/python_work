字符串的乐趣

1. 回⽂是指⼀个词或短语，它从前往后读和从后往前读都⼀样。例如，level、
kayak 和 pop 都是英⽂回⽂；如果忽略空格和标点符号，rats live on no
evil star 和 Was it a car or a cat I saw? 也是回⽂短语。编写⼀个名
为 is_palindrome 的函数，它接受⼀个字符串作为唯⼀参数，并返回布尔值：如
果输⼊是回⽂，返回 True，否则返回 False。您可以假设输⼊仅包含英⽂字⺟
（不区分⼤⼩写）、数字 0 到 9 和空格。判断回⽂时应忽略空格和⼤⼩写，因此
tacocat 和 T A C O cat 都应被认为是回⽂。

2. 如果⼀个词的字⺟从左到右按字⺟顺序排列，就称它是“字⺟顺序的”；允许字⺟重
复。例如，adder 和 beet 是字⺟顺序的，dog 和 cat 则不是。编写⼀个名为
is_abecedarian 的函数，它只接受⼀个字符串参数。如果参数符合上述要求，
返回 True，否则返回 False。您可以假设输⼊仅由字⺟和空格组成。判断时应忽
略空格，因此 abcd efgh xyz 被认为是字⺟顺序的。

3. 编写⼀个名为 remove_vowels 的函数，它接受⼀个字符串作为唯⼀参数，并返回
去掉所有元⾳后的字符串。本题的元⾳是字⺟ a、e、i、o、u。例如，
remove_vowels('cat') 应返回 'ct'，remove_vowels('audio') 应返回
'd'。函数也应正确处理全部由元⾳组成的字符串。

```python
def is_palindrome(s):
    # 去掉空格，全部转为小写
    clean = s.replace(" ", "").lower()
    # 判断和反转后的字符串是否相等
    return clean == clean[::-1]

def is_abecedarian(s):
    # 去掉空格，全部转小写
    clean = s.replace(" ", "").lower()
    # 依次检查每个字符是否不小于前一个
    for i in range(1, len(clean)):
        if clean[i] < clean[i-1]:
            return False
    return True

def remove_vowels(s):
    vowels = {'a', 'e', 'i', 'o', 'u'}
    result = ""
    for char in s:
        if char.lower() not in vowels:
            result += char
    return result

# ========== 测试 ==========
if __name__ == "__main__":
    # 测试回文
    print("测试回文")
    print(is_palindrome("tacocat"))         # True
    print(is_palindrome("T A C O cat"))     # True
    print(is_palindrome("hello"))           # False

    # 测试字母顺序
    print("\n测试字母顺序")
    print(is_abecedarian("adder"))          # True
    print(is_abecedarian("abcd efgh xyz"))  # True
    print(is_abecedarian("dog"))            # False

    # 测试去元音
    print("\n测试去元音")
    print(remove_vowels('cat'))             # ct
    print(remove_vowels('audio'))           # d
    print(remove_vowels('aeiou'))          # 空字符串
```

列表的乐趣

1. 编写⼀个名为 list_reverse 的函数，它接受⼀个列表作为参数，并返回反转后
的列表。例如，输⼊ [1, 2, 3]，应返回 [3, 2, 1]。如果输⼊不是列表，函数
应引发适当的错误。

2. 编写⼀个名为 is_sorted 的函数，它接受⼀个序列 seq 作为唯⼀参数。如果序列
按升序排列，返回 True，否则返回 False。您可以假设 seq 是 Python 序列，且
所有元素都⽀持⽐较操作，因此⽆需检查输⼊类型。函数只应遍历序列⼀次；需要
多次遍历的低效解法不能获得满分。这个问题展⽰了 Python 动态类型的特点：只
要 seq ⽀持索引，且元素⽀持⽐较，就⽆需预先知道序列的具体类型。

```python
def list_reverse(lst):
    if not isinstance(lst, list):
        raise TypeError("输入必须是列表类型")
    return lst[::-1]

def is_sorted(seq):
    for i in range(len(seq) - 1):
        if seq[i] > seq[i + 1]:
            return False
    return True

# ========== 测试 ==========
if __name__ == "__main__":
    print("==== list_reverse 测试 ====")
    print(list_reverse([1, 2, 3]))
    print(list_reverse(["a", "b", "c"]))

    print("\n==== is_sorted 测试 ====")
    print(is_sorted([1, 2, 3, 4]))
    print(is_sorted([1, 3, 2]))
    print(is_sorted((2, 2, 5, 7)))
    print(is_sorted([]))
    print(is_sorted([99]))
```

作为向量的元组

1. 实现⼀个名为 vec_scalar_mult 的函数。它接受两个参数：⼀个由整数或浮点
数组成的元组 t，以及⼀个整数或浮点数 s。函数返回与 t ⻓度相同的元组，每个
元素等于 t 的对应元素乘以 s。函数应检查输⼊类型；类型不合适时，引发带有适
当错误消息的 TypeError。整数标量和含有整数元素的元组都是有效输⼊。

```python
def vec_scalar_mult(t, s):
    # 检查 t 是否为元组
    if not isinstance(t, tuple):
        raise TypeError("第一个参数必须是元组")
    # 检查 s 是否是 int / float
    if not isinstance(s, (int, float)):
        raise TypeError("标量 s 必须是整数或浮点数")
    # 检查元组内每个元素是 int/float
    for elem in t:
        if not isinstance(elem, (int, float)):
            raise TypeError("向量元组中的元素必须是整数或浮点数")
    return tuple(x * s for x in t)

t1 = (1, 2, 3)
print(vec_scalar_mult(t1, 2))  # (2, 4, 6)
t2 = (1.5, -2, 4.0)
print(vec_scalar_mult(t2, -3))
t3 = ()
print(vec_scalar_mult(t3, 5))
try:
    vec_scalar_mult([1,2,3], 2)
except TypeError as e:
    print(f"vec_scalar_mult([1,2,3],2) 报错：{e}")
```

2. 实现⼀个名为 vec_inner_product 的函数。它接受两个由整数或浮点数组成的元组，返回这两个向量内积对应的浮点数。向量$x,y \in \mathbb{R}^d$ 的内积是$\sum_{j=1}^{d} x_j y_j$
。若任⼀输⼊不是元组，应引发 TypeError；若两个元组的⻓度不⼀致，应引发
ValueError。

```python
def vec_inner_product(x, y):
    if not isinstance(x, tuple) or not isinstance(y, tuple):
        raise TypeError("输入向量必须是元组")
    if len(x) != len(y):
        raise ValueError("两个向量长度必须相等")
    total = 0.0
    for xi, yi in zip(x, y):
        if not (isinstance(xi, (int, float)) and isinstance(yi, (int, float))):
            raise TypeError("向量元素必须是整数或浮点数")
        total += xi * yi
    return total

v1 = (1, 2, 3)
v2 = (4, 5, 6)
print(vec_inner_product(v1, v2))  # 1*4 + 2*5 +3*6 = 32.0
try:
    vec_inner_product((1,2), (3,4,5))
except ValueError as e:
    print(f"vec_inner_product((1,2),(3,4,5)) 报错：{e}")
```

3. 将上述表⽰⽅法扩展到矩阵。⽤⼀个元组的元组表⽰$m*n$ 矩阵 my_mx：外层元组⻓度为 ，每个内层元组⻓度为 ，表⽰矩阵的⼀⾏。编写⼀个名为check_valid_mx 的函数，它接受⼀个参数；若该参数是有效矩阵则返回 True，否则返回 False。有效矩阵满⾜：
外层元组的每个元素本⾝都是元组；
所有内层元组⻓度相同；
每个内层元组的每个元素都是整数或浮点数。

```python
def check_valid_mx(mx):
    # 外层必须是元组
    if not isinstance(mx, tuple):
        return False
    # 空矩阵也算合法
    if len(mx) == 0:
        return True
    row_len = len(mx[0])
    for row in mx:
        # 每一行必须是元组
        if not isinstance(row, tuple):
            return False
        # 每行长度一致
        if len(row) != row_len:
            return False
        # 每个元素是 int/float
        for elem in row:
            if not isinstance(elem, (int, float)):
                return False
    return True

mat1 = ((1, 2), (3, 4))
print(check_valid_mx(mat1))  # True
mat2 = ((1.1, 2.2, 3), (4, 5, 6.6))
print(check_valid_mx(mat2))
mat_empty = ()
print(check_valid_mx(mat_empty))
mat_bad1 = ((1,2), (3,))
print(check_valid_mx(mat_bad1))
mat_bad2 = ([1,2], (3,4))
print(check_valid_mx(mat_bad2))
```

4. 编写⼀个名为 mx_vec_mult 的函数，它接受⼀个矩阵（元组的元组）和⼀个向量
（元组），返回向量与矩阵相乘得到的数值元组。函数check_valid_mx应检查参
数是否合理；对于不合理的输⼊，应引发适当的错误。

```python
def mx_vec_mult(mx, v):
    # 先校验矩阵
    if not check_valid_mx(mx):
        raise TypeError("输入不是合法矩阵（元组的元组）")
    # 校验向量
    if not isinstance(v, tuple):
        raise TypeError("向量必须是元组")
    for elem in v:
        if not isinstance(elem, (int, float)):
            raise TypeError("向量元素必须是整数或浮点数")
    # 矩阵列数 = 向量长度
    if len(mx) > 0 and len(mx[0]) != len(v):
        raise ValueError(f"矩阵的列数 {len(mx[0])} 必须等于向量维度 {len(v)}")

    res = []
    for row in mx:
        dot = vec_inner_product(row, v)
        res.append(dot)
    return tuple(res)

print(mx_vec_mult(mat1, (5, 6)))
try:
    mx_vec_mult( ((1,2),(3,4)), (1,) )
except ValueError as e:
    print(f"mx_vec_mult(((1,2),(3,4)), (1,)) 报错：{e}")
```

更多向量的乐趣

1. 定义⼀个名为 Vector 的类。每个向量都有⼀个维度（⾮负整数）和⼀个元素列表或元组。初始化函数的第⼀个参数是维度，第⼆个参数是由整数或浮点数组成的列表或元组，表⽰向量的元素。如果只提供维度⽽不提供元素，请选择合理的默认⾏为。若维度的类型不正确或为负数，应抛出合适的异常；若维度与元素个数不⼀致，也应抛出异常。

2. 维度和元素应是类属性还是实例属性？说明选择的理由。

维度和元素应该是实例属性。
理由：每一个 Vector 向量对象可以拥有互不相同的维度与元素。例如可以同时存在 2 维向量、3 维向量，每个向量的元素也各自独立。
类属性属于整个类、所有对象共享同一份数据，不适合存放每个向量独有的信息；而实例属性是每个实例单独拥有，符合向量的语义。

3. 实现向量对象的⽐较运算符，包括相等、⼩于、⼩于或等于、⼤于等。两个
Vector 对象的坐标相同，则视为相等；否则按 Python 元组的字典序⽐较，即先
⽐较第⼀个坐标，再⽐较第⼆个坐标，依此类推。例如，⼆维向量 (2, 4) ⼩于
(2, 5)。⽐较维度不同的两个向量时，应抛出异常。

4. 实现 Vector.dot ⽅法，接收另⼀个 Vector 对象，返回两向量的内积。若参数
类型不正确或维度不⼀致，应抛出合适的异常。

5. 实现 Vector 的乘法运算符。标量左乘或右乘（如 $2*v$ 和 $v*2$）应返回⼀个各元
素都乘以该标量的新 Vector 对象。当两个 Vector 对象相乘$（v*w）$时，对应位置的元素逐⼀相乘，结果仍是⼀个新 Vector 对象；例如，$Vector(2, (2, 3))* Vector(2, (4, 5))$ 的元素为 (8, 15)。若两个向量的维度不⼀致，应抛出
合适的异常。

```python
class Vector:
    def __init__(self, dim, elements=None):
        """
        构造Vector向量类
        :param dim: 维度，非负整数
        :param elements: 列表/元组，int/float组成；不传则默认全0
        """
        # 检查维度类型与取值
        if not isinstance(dim, int):
            raise TypeError("维度必须是整数")
        if dim < 0:
            raise ValueError("维度不能为负数")

        if elements is None:
            # 只给维度、不给元素：默认全部为0
            elements = [0.0] * dim
        else:
            # 转为元组统一存储
            elements = tuple(elements)
            # 检查元素都是数字
            for e in elements:
                if not isinstance(e, (int, float)):
                    raise TypeError("向量元素只能是整数或浮点数")
            # 维度和元素数量匹配校验
            if len(elements) != dim:
                raise ValueError(f"维度 {dim} 和元素数量 {len(elements)} 不一致")

        # 维度dim、元素elements 作为【实例属性】
        self.dim = dim
        self.elements = elements

    def __repr__(self):
        return f"Vector({self.dim}, {self.elements})"

    # ========= 3. 比较运算符：==, <, <=, >, >= =========
    def __eq__(self, other):
        if not isinstance(other, Vector):
            return False
        if self.dim != other.dim:
            raise ValueError("不能比较维度不同的向量")
        return self.elements == other.elements

    def __lt__(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能和Vector对象比较")
        if self.dim != other.dim:
            raise ValueError("不能比较维度不同的向量")
        return self.elements < other.elements

    def __le__(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能和Vector对象比较")
        if self.dim != other.dim:
            raise ValueError("不能比较维度不同的向量")
        return self.elements <= other.elements

    def __gt__(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能和Vector对象比较")
        if self.dim != other.dim:
            raise ValueError("不能比较维度不同的向量")
        return self.elements > other.elements

    def __ge__(self, other):
        if not isinstance(other, Vector):
            raise TypeError("只能和Vector对象比较")
        if self.dim != other.dim:
            raise ValueError("不能比较维度不同的向量")
        return self.elements >= other.elements

    # ========= 4. dot 内积方法 =========
    def dot(self, other):
        if not isinstance(other, Vector):
            raise TypeError("内积的参数必须是Vector实例")
        if self.dim != other.dim:
            raise ValueError("计算内积要求两个向量维度相同")
        total = 0.0
        for a, b in zip(self.elements, other.elements):
            total += a * b
        return total

    # ========= 5. 乘法运算符 * =========
    def __mul__(self, other):
        # v * s  向量 * 标量
        if isinstance(other, (int, float)):
            new_elems = tuple(x * other for x in self.elements)
            return Vector(self.dim, new_elems)
        # v * v2 向量逐元素相乘（哈达玛积）
        elif isinstance(other, Vector):
            if self.dim != other.dim:
                raise ValueError("逐元素相乘要求两个向量维度一致")
            new_elems = tuple(a * b for a, b in zip(self.elements, other.elements))
            return Vector(self.dim, new_elems)
        else:
            raise TypeError("乘法仅支持标量(int/float)或者Vector对象")

    # 2 * v 标量左乘
    def __rmul__(self, other):
        return self * other


# ===================== 测试代码 =====================
if __name__ == "__main__":
    print("==== 初始化测试 ====")
    v0 = Vector(3)
    print(f"v0 = {v0}")
    v1 = Vector(2, (2, 4))
    v2 = Vector(2, (2, 5))
    print(f"v1 = {v1}, v2 = {v2}")

    print("\n==== 比较运算测试 ====")
    print(f"v1 == v2 ? {v1 == v2}")
    print(f"v1 <  v2 ? {v1 < v2}")
    print(f"v1 <= v2 ? {v1 <= v2}")
    print(f"v1 >  v2 ? {v1 > v2}")

    print("\n==== dot 内积测试 ====")
    va = Vector(3, (1, 2, 3))
    vb = Vector(3, (4, 5, 6))
    print(f"va.dot(vb) = {va.dot(vb)}")

    print("\n==== 乘法测试：标量、向量逐元素乘 ====")
    v3 = Vector(2, (2, 3))
    v4 = Vector(2, (4, 5))
    print(f"v3 * 2 = {v3 * 2}")
    print(f"3 * v3 = {3 * v3}")
    print(f"v3 * v4 = {v3 * v4}")
```
