1. 编写score_level(score)函数：90〜100分返回“优秀”，80〜不⾜90分返回“良好”，60〜不⾜80分返回“及格”，0〜不⾜60分返回“不及格”。对于超出0〜100范围的数值，抛出ValueError。

```python
def score_level(score):
    if not (0 <= score <= 100):
        raise ValueError("分数必须在0~100之间")
    if 90 <= score <= 100:
        return "优秀"
    elif 80 <= score < 90:
        return "良好"
    elif 60 <= score < 80:
        return "及格"
    else:  # 0 <= score <60
        return "不及格"

# 测试
print(score_level(95))
print(score_level(85))
print(score_level(70))
print(score_level(50))
try:
    score_level(105)
except ValueError as e:
    print("ValueError:", e)
```

3. 编写递归版和迭代版阶乘函数，约定0! = 1，输⼊为⾮负整数。⽐较n = 0、1、5、10时的结果。

```python
# 迭代版
def fact_iter(n):
    if not isinstance(n, int) or n < 0:
        raise ValueError("输入必须是非负整数")
    res = 1
    for i in range(2, n+1):
        res *= i
    return res

# 递归版
def fact_recur(n):
    if not isinstance(n, int) or n < 0:
        raise ValueError("输入必须是非负整数")
    if n == 0 or n == 1:
        return 1
    return n * fact_recur(n - 1)

for n in [0,1,5,10]:
    print(f"n={n}, 迭代阶乘={fact_iter(n)}, 递归阶乘={fact_recur(n)}")
```

4. 编写summary(values)函数，接收⾮空数值列表，以元组返回最⼩值、最⼤值和平均值，并通过元组解包接收结果。

```python
def summary(values):
    if len(values) == 0:
        raise ValueError("列表不能为空")
    min_v = min(values)
    max_v = max(values)
    avg_v = sum(values) / len(values)
    return (min_v, max_v, avg_v)

lst = [10, 20, 30, 40, 50]
min_val, max_val, avg_val = summary(lst) # 元组解包
print(f"最小值={min_val}, 最大值={max_val}, 平均值={avg_val:.2f}")
```

3. 编写parse_student(line)函数：将成绩转换为整数，检查字段数量、姓名是否为空及成绩是否在0〜100范围内。⽆效记录抛出ValueError。

```python
def parse_student(line):
    line = line.strip()
    parts = line.split(",")
    # 检查字段数量，必须2列
    if len(parts) != 2:
        raise ValueError("字段数量错误，必须是 name,score 两列")
    name, score_str = parts
    name = name.strip()
    if len(name) == 0:
        raise ValueError("姓名不能为空")
    # 转为整数
    try:
        score = int(score_str.strip())
    except:
        raise ValueError("成绩无法转为整数")
    # 成绩范围0~100
    if not (0 <= score <= 100):
        raise ValueError("成绩超出0~100范围")
    return {"name": name, "score": score}
```

4. 按⾏处理⽂件，捕获⽆效记录的异常，记录⾏号与错误原因，并继续处理后续记录。有效记录保存为字典组成的列表。

```python
valid_list = []
error_log = []
with open(stu_txt_path, "r", encoding="utf-8") as f:
    lines = f.readlines()

for line_no, raw_line in enumerate(lines, start=1):
    line = raw_line.strip()
    if line_no == 1:
        continue  # 跳过表头
    if not line:
        continue
    try:
        stu_dict = parse_student(line)
        valid_list.append(stu_dict)
    except ValueError as err:
        error_log.append({"行号": line_no, "错误": str(err)})

print("有效记录：")
for s in valid_list:
    print(s)
print("\n无效记录日志：")
for err in error_log:
    print(err)

# 统计
valid_cnt = len(valid_list)
invalid_cnt = len(error_log)
if valid_cnt > 0:
    avg_score = sum(s["score"] for s in valid_list) / valid_cnt
    pass_cnt = sum(1 for s in valid_list if s["score"] >= 60)
    pass_rate = pass_cnt / valid_cnt
else:
    avg_score = 0
    pass_rate = 0
```

实验5：类、继承与对象复制

1. 定义Student类，在__init__()中初始化name和score属性，并实现is_passed()
和__str__()⽅法。

```python
import copy

class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        # >=60 返回True及格
        return self.score >= 60

    def __str__(self):
        return f"Student(name={self.name}, score={self.score})"
```

2. 为实验4中的每条有效记录创建⼀个Student对象，输出对象信息和及格状态。

```python
# 实验4有效记录
valid_records = [
    {"name": "Alice", "score": 85},
    {"name": "Bob", "score": 58},
    {"name": "Carol", "score": 92},
    {"name": "David", "score": 76},
    {"name": "Eva", "score": 60},
    {"name": "Frank", "score": 89}
]

stu_obj_list = []
for rec in valid_records:
    s = Student(rec["name"], rec["score"])
    stu_obj_list.append(s)
    print(s, "及格" if s.is_passed() else "不及格")
```

3. 定义继承⾃Student的GraduateStudent类，增加research_topic属性，通过
super().__init()__初始化⽗类属性，并重写__str__()。

```python
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        # 调用父类构造函数
        super().__init__(name, score)
        self.research_topic = research_topic

    # 重写__str__
    def __str__(self):
        return f"GraduateStudent(name={self.name}, score={self.score}, research_topic={self.research_topic})"

# 测试
g_stu = GraduateStudent("LiMing", 88, "机器学习")
print(g_stu, "及格" if g_stu.is_passed() else "不及格")
```
